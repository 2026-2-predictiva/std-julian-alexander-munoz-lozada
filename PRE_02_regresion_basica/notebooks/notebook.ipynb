```python
import json
import pickle
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import sklearn
from joblib import Memory

base = next(
    p for r in [Path.cwd(), *Path.cwd().parents]
    for p in [r, r / "PRE_02_regresion_basica"]
    if p.name == "PRE_02_regresion_basica" and (p / "notebooks").is_dir()
)
submission = base / "submission"
submission.mkdir(exist_ok=True)
cache = Memory(base / "temp/cache", verbose=0)

dataset = pd.read_csv(base / "data/auto_mpg.csv")
dataset.head()
```

```python
dataset.shape
```

```python
dataset.isna().sum()
```

```python
dataset = dataset.dropna().copy()
dataset.isna().sum()
```

```python
dataset.Origin.value_counts()
```

```python
dataset["Origin"] = dataset["Origin"].map(
    {   
        1: "USA", 
        2: "Europe", 
        3: "Japan",
    },
)
dataset.head(20)
```

```python
dataset = pd.get_dummies(dataset, columns=["Origin"], prefix="", prefix_sep="")
dataset.head()
```

```python
train_dataset = dataset.sample(frac=0.8, random_state=0)
test_dataset = dataset.drop(train_dataset.index)
```

```python
import seaborn as sns

sns.pairplot(
    train_dataset[["MPG", "Cylinders", "Displacement", "Weight"]], diag_kind="kde"
)
plt.show()
```

```python
train_dataset.describe().transpose()
```

```python
train_features = train_dataset.copy()
train_labels = train_features.pop("MPG")

test_features = test_dataset.copy()
test_labels = test_features.pop("MPG")
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
```

```python
pd.DataFrame(
    data=scaler.fit_transform(train_features),
    columns=train_features.columns,
).describe().transpose()[["mean", "std"]]
```

```python
horsepower_scaler = StandardScaler()

train_horsepower = train_features[["Horsepower"]]
test_horsepower = test_features[["Horsepower"]]

horsepower_scaler.fit(train_horsepower)

standarized_train_horsepower = horsepower_scaler.transform(train_horsepower)
standarized_test_horsepower = horsepower_scaler.transform(test_horsepower)
```

```python
from sklearn.linear_model import LinearRegression

horsepower_model = LinearRegression()
horsepower_model.fit(standarized_train_horsepower, train_labels)
```

```python
horsepower_model.intercept_
```

```python
horsepower_model.coef_
```

```python
import numpy as np

x = pd.DataFrame({"Horsepower": np.linspace(0, 250, 251)})
x.head()
```

```python
scaled_x = horsepower_scaler.transform(x)
y = horsepower_model.predict(scaled_x)
y[:5]
```

```python
import matplotlib.pyplot as plt


def plot_horsepower(x, y):
    plt.scatter(train_features["Horsepower"], train_labels, label="Data")
    plt.plot(x, y, color="k", label="Predictions")
    plt.xlabel("Horsepower")
    plt.ylabel("MPG")
    plt.legend()
    
    
    
    
    
plot_horsepower(x, y)
plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import mean_squared_error

test_results = {}

y_pred = horsepower_model.predict(standarized_test_horsepower)

test_results["horsepower_model"] = mean_squared_error(
    y_true=test_labels,
    y_pred=y_pred,
)

test_results
```

```python
from sklearn.neural_network import MLPRegressor

config = {"hidden_layer_sizes": [16], "solver": "lbfgs", "alpha": 1.0,
          "max_iter": 10000, "random_state": 42, "split_seed": 0,
          "train_fraction": 0.8, "sklearn": sklearn.__version__}
(submission / "config.json").write_text(json.dumps(config, indent=2))

@cache.cache
def entrenar(features, labels, config):
    scaler = StandardScaler()
    scaled = scaler.fit_transform(features)
    modelo = MLPRegressor(
        hidden_layer_sizes=tuple(config["hidden_layer_sizes"]),
        solver=config["solver"], alpha=config["alpha"],
        max_iter=config["max_iter"], random_state=config["random_state"],
    )
    modelo.fit(scaled, labels)
    return scaler, modelo

features_scaler, mlp = entrenar(train_features, train_labels, config)
```

```python
y_pred = mlp.predict(features_scaler.transform(test_features))
test_results["mlp"] = mean_squared_error(test_labels, y_pred)
pd.Series(test_results, name="MSE de prueba")
```

```python
plt.figure(figsize=(6, 4))
plt.scatter(test_labels, y_pred, alpha=0.7)
limites = [min(test_labels.min(), y_pred.min()), max(test_labels.max(), y_pred.max())]
plt.plot(limites, limites, "k--")
plt.xlabel("MPG real")
plt.ylabel("MPG estimado")
plt.tight_layout()
plt.show()
```

```python
for nombre, objeto in {"mlp.pkl": mlp, "features_scaler.pkl": features_scaler}.items():
    with (submission / nombre).open("wb") as archivo:
        pickle.dump(objeto, archivo)
pd.Series(test_results, name="mse_prueba").to_csv(submission / "metricas.csv")
```
