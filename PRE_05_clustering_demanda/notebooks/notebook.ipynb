```python
import json
from pathlib import Path

import pandas as pd
import matplotlib.pyplot as plt
import sklearn
from joblib import Memory

base = next(
    p for r in [Path.cwd(), *Path.cwd().parents]
    for p in [r, r / "PRE_05_clustering_demanda"]
    if p.name == "PRE_05_clustering_demanda" and (p / "data").is_dir()
)
submission = base / "submission"
submission.mkdir(exist_ok=True)
cache = Memory(base / "temp" / "cache", verbose=0)

data = pd.read_csv(base / "data/demanda_comercial.csv.zip")
data.head()
```

```python
data_to_plot = data.copy()
data_to_plot = data_to_plot.melt(
    id_vars=["Fecha"], var_name="hora", value_name="demanda"
)
data_to_plot = data_to_plot.sort_values(by=["Fecha", "hora"])
data_to_plot = data_to_plot.reset_index(drop=True)
data_to_plot = data_to_plot.head(500)
data_to_plot.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(data_to_plot["demanda"])
plt.title("Demanda comercial")
plt.xlabel("Tiempo")
plt.ylabel("Demanda")
plt.grid()
plt.tight_layout()
plt.savefig(submission / "demanda-comercial.png")
plt.show()
```

```python
data = data.dropna()
data = data.drop_duplicates()
data = data.set_index("Fecha")

data.head()
```

```python
plt.figure(figsize=(7, 4))
plt.plot(data.loc["2017-06-05", :], ".-", color="tab:blue")
plt.plot(data.loc["2018-06-05", :], ".-", color="tab:orange")
plt.plot(data.loc["2022-06-03", :], ".-", color="tab:green")
plt.xticks(rotation=90)
plt.show()
```

```python
data = data.div(data.max(axis=1).replace(0, float("nan")), axis=0).dropna()
data.head(10)
```

```python
plt.figure(figsize=(7, 4))
plt.plot(data.loc["2017-06-05", :], ".-", color="tab:blue")
plt.plot(data.loc["2018-06-05", :], ".-", color="tab:orange")
plt.plot(data.loc["2022-06-03", :], ".-", color="tab:green")
plt.xticks(rotation=90)
plt.tight_layout()
plt.savefig(submission / "demanda-comercial-patrones-ejemplo.png")
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

config = {"n_clusters": 2, "n_init": 10, "random_state": 42,
          "candidatos": list(range(2, 10)), "normalizacion": "maximo_diario",
          "sklearn": sklearn.__version__}
(submission / "config.json").write_text(json.dumps(config, indent=2))

@cache.cache
def agrupar(data, n_clusters, n_init, random_state, version):
    modelo = KMeans(n_clusters=n_clusters, n_init=n_init, random_state=random_state)
    modelo.fit(data)
    score = silhouette_score(data, modelo.labels_)
    return modelo, score

resultados = {
    n: agrupar(data, n, config["n_init"], config["random_state"], config["sklearn"])
    for n in config["candidatos"]
}
scores = [resultados[n][1] for n in config["candidatos"]]

plt.figure(figsize=(5, 4))
plt.plot(config["candidatos"], scores, marker="o", color="tab:blue")
plt.xlabel("Numero de clusters")
plt.ylabel("Silhouette")
plt.grid()
plt.tight_layout()
plt.show()
```

```python
kmeans = resultados[config["n_clusters"]][0]

plt.figure(figsize=(7, 4))
for n, perfil in enumerate(kmeans.cluster_centers_):
    plt.plot(data.columns, perfil, ".-", label=f"Cluster {n}")
plt.xlabel("Hora")
plt.ylabel("Demanda normalizada")
plt.xticks(rotation=90)
plt.legend()
plt.tight_layout()
plt.savefig(submission / "demanda-comercial-perfiles.png")
plt.show()
```

```python
data = data.assign(cluster=kmeans.labels_)
data = data.assign(day=pd.to_datetime(data.index).day_of_week)
data[["cluster", "day"]]
```

```python
days_cluster_0 = data.loc[data.cluster == 0, "day"].value_counts().reindex(range(7), fill_value=0)
days_cluster_0
```

```python
days_cluster_1 = data.loc[data.cluster == 1, "day"].value_counts().reindex(range(7), fill_value=0)
days_cluster_1
```

```python
resumen = pd.DataFrame({
    "days": range(7),
    "cluster 0": days_cluster_0,
    "cluster 1": days_cluster_1,
})
resumen.to_csv(submission / "demanda-comercial-dias.csv", index=False)
resumen
```
